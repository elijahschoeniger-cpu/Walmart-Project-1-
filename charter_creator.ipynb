```python
# Run this once the charter above has no unfilled items. It writes CHARTER.md next to this notebook.
def charter_markdown(t):
    lines = [f"# Team charter: {t['name']}", '', '## Members']
    lines += [f"- {m['name']} (GitHub: {m['github']})" for m in t['members']]
    lines += ['', '## Goals']
    lines += [f"- **{k.replace('_', ' ')}:** {v}" for k, v in t['goals'].items()]
    lines += ['', '## Roles', '', '| Role | Owner | Backup |', '|---|---|---|']
    lines += [f"| {r} | {v['owner']} | {v['backup']} |" for r, v in t['roles'].items()]
    lines += ['', '## Question owners', '', '| Question | Owner | Reviewer |', '|---|---|---|']
    lines += [f"| {q} | {v['owner']} | {v['reviewer']} |" for q, v in t['question_owners'].items()]
    lines += ['', '## How we work']
    for k in ('repo_url', 'notebook_owner_this_week', 'meeting_time',
              'channel', 'how_we_decide', 'if_someone_goes_quiet'):
        lines.append(f"- **{k.replace('_', ' ')}:** {t[k]}")
    return '\n'.join(lines) + '\n'

if unfilled:
    print('Finish the charter above first, then run this cell.')
else:
    with open('CHARTER.md', 'w') as f:
        f.write(charter_markdown(team))
    print('wrote CHARTER.md -- move it to the root of your repo and commit it')
```
